```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc=QuantumCircuit(2)
qc.h(0)
qc.x(1)
qc.cx(0,1)
qc.measure_all()

sim=AerSimulator()
s=sim.run(qc,shots=1000)
m=s.result()
d=m.get_counts()
k=list(d.keys())[0]

print(d)
print(k)

qc.draw("mpl")
```
