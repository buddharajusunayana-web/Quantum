```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc=QuantumCircuit(2,0)
qc.h(0)
qc.cx(0,1)

qc.id(0)
qc.cx(0,1)
qc.h(0)
qc.measure_all()
sim=AerSimulator()
a=sim.run(qc,shots=1000)
b=a.result()
c=b.get_counts()
d=list(c.keys())[0]

print(d)
qc.draw("mpl")
```
