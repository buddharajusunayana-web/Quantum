```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc=QuantumCircuit(2,0)
qc.x(0)
#qc.x(1)
qc.cx(0,1)
qc.measure_all()

for i in range(10):
  sim=AerSimulator()
  job=sim.run(qc,shots=1000)
  m=job.result()
  s=m.get_counts()
  d=list(s.keys())[0]
  print(s)
  print(d)

qc.draw("mpl")
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc=QuantumCircuit(2)
qc.x(0)
qc.iswap(0,1)
qc.measure_all()

compiled_qc = transpile(qc, sim)

sim=AerSimulator()
job=sim.run(compiled_qc,shots=10000)
s=job.result()
m=s.get_counts()
d=list(m.keys())[0]
print(m)
print(d)

qc.draw("mpl")
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import numpy as np


qc=QuantumCircuit(2)
qc.h(0)
qc.p(np.pi/4,0)
qc.h(0)
qc.measure_all()

compiled_qc = transpile(qc, sim)

sim=AerSimulator()
job=sim.run(compiled_qc,shots=100)
s=job.result()
m=s.get_counts()
d=list(m.keys())[0]
print(m)
print(d)

qc.draw("mpl")
```
