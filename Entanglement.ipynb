```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc=QuantumCircuit(2)
qc.h(0)
qc.x(1)
qc.cx(0,1)
qc.measure_all()

sim=AerSimulator()
x=sim.run(qc,shots=1000)
s=x.result()
m=s.get_counts()
d=list(m.keys())[0]
print(m)
print(d)

qc.draw("mpl")
```
