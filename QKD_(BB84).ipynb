```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

n=int(input("Enter N:"))
alice_bits=np.random.randint(0,2,n)
alice_basis=np.random.randint(0,2,n)

print("Alice Bits:",alice_bits)
print("Alice_Basis:",alice_basis)

bob_basis=np.random.randint(0,2,n)
print("Bob_Basis:",bob_basis)

qc=QuantumCircuit(n,n)

for i in range(n):
  if alice_bits[i]==1: # is alice bit is 0 leave it as it is as by default qubits are assigned 0 if not 1
    qc.x(i)

  if alice_basis[i]==1: # 0 is considered as standard basis and 1 for hadamard basis
    qc.h(i)

  if bob_basis[i]==1:
    qc.h(i)

qc.measure(range(n),range(n))

print(qc)

sim=AerSimulator()
runq=sim.run(qc,shots=1000)
res=runq.result()
count=res.get_counts()
final=list(count.keys())[0]


print(count)
print(final)

bob_bits=np.array([int(bit) for bit in final[::-1]])
print(bob_bits)

match=alice_bits==bob_bits
print(match)

alice_key=alice_bits[match]
bob_key=bob_bits[match]

print(alice_key)
print(bob_key)

non_match=np.sum(alice_bits!=bob_bits)
print(non_match)

qber=(non_match/n)*100

print(qber)

```
